```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import FAISS
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

load_dotenv()
```

```python
loader = PyPDFLoader(r"data\ResearchBuddy---Academic-Research-Agent-prd-XWwr8.pdf")
docs = loader.load()
```

```python
len(docs)
```

```python
print(docs)
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=200)
chunks = splitter.split_documents(docs)
```

```python
len(chunks)
```

```python
print(chunks)
```

```python
embedding_model = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2")
embedding_model
```

```python
vector_store = FAISS.from_documents(chunks,embedding_model)
vector_store
```

```python
vector_store.similarity_search("What is ResearchBuddy?")
```

```python
retriever = vector_store.as_retriever()
retriever
```

```python
retriever.invoke("What is ResearchBuddy?")
```

```python
prompt_template = '''

Answer the question based on the context provided below.
If the context do not contain sufficient informatioin, respond with:
"I do not have enough information about this."

Context: {context}
Question: {question}
Answer:
'''
```

```python
prompt = PromptTemplate.from_template(template=prompt_template)
prompt
```

```python
model = ChatGroq(model="openai/gpt-oss-120b")
model
```

```python
model.invoke("Hi")
```

```python
parser = StrOutputParser()
parser
```

```python
rag_chain = prompt | model | parser
rag_chain
```

```python
def format_docs(query):
    context = ""
    docs = retriever.invoke(query)
    for doc in docs:
        context+= doc.page_content + "\n\n"

    return context
```

```python
query = "tell me about researchbuddy??"

response = rag_chain.invoke({"question":query,"context":format_docs(query)})
response
```
